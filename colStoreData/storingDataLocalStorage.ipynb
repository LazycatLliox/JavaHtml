```python
var name = prompt("Enter your first name:");

// This can be stored as a key : value pair in local storage
```

```python
// The javascript keyword to store data
localStorage.setItem("key", "value");
```

```python
/* retrive and store the data in the variable */
var fName = prompt("Enter your first name:");

localStorage.setItem("user", fName);

// To retrieve the data the data that was stored in the local storage
//var retrieveName = localStorage.getItem("user");
var retrieveName = localStorage.user;
```

### Null Data Type

<font color= "red">null</font> - an absense of data, no data exists

The null data type is return when use getItem(key) and the dat dose not exist, however if use getItem.key and the data does not exist it will return <b>Undefined</b>

Where it is Null or Undefined - the expression evaluates to a <b>False Response</b>

Both: 
if(localStorage.getItem("key") and if(localStorage.key))

will evaluate false if the key doesn't exist
